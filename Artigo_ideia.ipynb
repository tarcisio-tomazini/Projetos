## ATIVIDADES EXPERIMENTAIS PARA 1º SÉRIE DO ENSINO MÉDIO


__Experimentos práticos adequados para a 1ª série do Ensino Médio__

__Foco:__ _Mecânica, Cinemática e Dinâmica_

__Estudo da Queda Livre__

> Materiais: Esferas de diferentes massas (ou tamanhos), trena ou fita métrica, cronômetro e suporte universal.

__Movimento Retilíneo Uniforme e Variado no Plano Inclinado__

> Materiais: Calha ou plano inclinado, carrinho de dinâmica, cronômetro, fita métrica e pequenos blocos de madeira para ajuste de altura.

__Verificação da Força de Atrito__

> Materiais: Bloco de madeira com gancho, dinamômetro, massas aferidas (pesos) e superfícies com diferentes texturas (lixa, tecido, plástico, vidro).

__Experimento da Inércia (1ª Lei de Newton)__

> Materiais: Um copo com borda lisa, um cartão de papelão (ou postal) e uma moeda.

__Conservação da Energia Mecânica com Pêndulo Simples__

> Materiais: Suporte universal, fio de barbante, esfera com furo ou gancho, trena e transferidor.

__Cálculo da Potência Humana__

> Materiais: Trena, balança (para medir a massa dos alunos), cronômetro e uma escada de acesso livre.


## ATIVIDADES EXPERIMENTAIS PARA 2º SÉRIE DO ENSINO MÉDIO


__Experimentos práticos adequados para a 2ª série do Ensino Médio__

__Foco:__ _Termologia, Calorimetria, Termodinâmica e Propriedades Térmicas dos Materiais_

**Dilatação Térmica Linear dos Metais**
> Materiais:* Dilatômetro linear (ou aparelho de Gravesande), barras de diferentes metais (alumínio, ferro, cobre), lamparina a álcool (ou bico de Bunsen) e régua milimetrada ou paquímetro.


**Determinação do Calor Específico (Calorimetria)**
> *Materiais:* Calorímetro (pode ser improvisado com copos de isopor e tampa), termômetro, água, balança digital, bico de Bunsen, béquer e corpos de prova metálicos (chumbo, alumínio, ferro).


**Condução Térmica em Diferentes Materiais**
> *Materiais:* Dispositivo de condutividade térmica (ou hastes de diferentes materiais: cobre, alumínio, ferro, vidro), tachinhas, parafina (ou cera de vela) e fonte de calor (vela ou bico de Bunsen).


**Estudo das Transformações Gasosas (Lei de Boyle com Seringa)**
> *Materiais:* Seringa de grande volume (50 ml ou 60 ml) com o bico vedado, livros ou pequenos pesos aferidos, e suporte ou régua para medir o deslocamento do êmbolo.


**Demonstração de Motor Térmico (Transformação de Calor em Trabalho)**
> *Materiais:* Lata de alumínio limpa e furada, água, suporte universal, vela ou fonte de calor e pequena hélice de papel alumínio para captar o vapor.





Aqui está a lista de experimentos práticos adequados para a 3ª série do Ensino Médio, focados nos conteúdos de Eletromagnetismo, Circuitos Elétricos, Ondulatória e Óptica:

Verificação da Lei de Ohm e Circuitos Elétricos Simples

Materiais: Pilhas ou fonte de tensão ajustável, resistores de valores conhecidos, multímetro digital, protoboard (matriz de contato) e fios de ligação.

Mapeamento de Linhas de Campo Elétrico / Magnético

Materiais: Ímãs permanentes (em barra e em ferradura), limalha de ferro, placa de acrílico transparente, papel sulfite e bobinas de fio de cobre (para eletroímã).

Construção e Teste de um Eletroímã

Materiais: Um prego grande de ferro, fio de cobre esmaltado, pilha (AA ou D), clipes de papel (para testar a atração magnética) e lixa fina.

Refração da Luz e Lei de Snell (Óptica Geométrica)

Materiais: Caixa de luz laser (ou laser de ponteiro), bloco de acrílico ou vidro em forma de semicírculo ou prisma, transferidor e folha de papel branco.

Construção de uma Lupa / Formação de Imagens em Lentes

Materiais: Lentes convergentes e divergentes, suporte para lentes, fonte de luz (vela ou objeto iluminado) e anteparo (tela branca ou papelão).

Ondas Estacionárias em Cordas ou Tubos Sonoros (Acústica)

Materiais: Gerador de sinais de áudio (ou aplicativo de smartphone), gerador de ondas mecânicas com corda (ou gerador de vibração por alto-falante) e tubos de ensaio ou PVC para ressonância com água.
